```python
query = """
SELECT 
    m.title,
    m.release_date
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
WHERE g.genre_name = 'Action'
  AND m.release_year > 2015
ORDER BY m.release_date;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The query shows the Action movies released after 2015 and their release dates, helping identify recent Action movie releases.

```python
query = """
SELECT
    m.title,
    m.revenue,
    GROUP_CONCAT(g.genre_name ORDER BY g.genre_name SEPARATOR ', ') AS genres
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
WHERE m.revenue > 0
GROUP BY m.movie_id, m.title, m.revenue
ORDER BY m.revenue DESC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The query identifies the 10 highest-revenue movies in the dataset and their genres, showing which movies were the biggest box-office performers.

```python
query = """
SELECT
    g.genre_name,
    AVG(m.budget) AS average_budget,
    AVG(m.revenue) AS average_revenue
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
WHERE m.budget > 0
  AND m.revenue > 0
GROUP BY g.genre_id, g.genre_name
ORDER BY average_revenue DESC;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The results compare the average budget and average revenue of each genre, showing which genres generate the highest average revenue among movies with known budget and revenue values.

```python
query = """
SELECT
    person_id,
    actor_name,
    COUNT(DISTINCT movie_id) AS movie_appearances
FROM cast
GROUP BY person_id, actor_name
ORDER BY movie_appearances DESC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The query identifies the 10 actors who appeared in the most movies, showing the most frequently featured actors in the dataset.

```python
query = """
SELECT
    person_id,
    person_name,
    COUNT(DISTINCT movie_id) AS movie_count
FROM crew
WHERE job = 'Director'
GROUP BY person_id, person_name
HAVING COUNT(DISTINCT movie_id) > 3
ORDER BY movie_count DESC;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The query identifies directors who have directed more than three movies, highlighting the most active and experienced directors in the dataset.

```python
query = """
SELECT
    keyword_name,
    COUNT(DISTINCT movie_id) AS movie_count
FROM movie_keywords
GROUP BY keyword_id, keyword_name
ORDER BY movie_count DESC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The query identifies the 10 keywords that appear in the most movies, showing the most common themes and topics in the dataset.

```python
query = """
SELECT
    title,
    budget,
    revenue
FROM movies
WHERE budget > (
    SELECT AVG(budget)
    FROM movies
    WHERE budget > 0
)
AND revenue < (
    SELECT AVG(revenue)
    FROM movies
    WHERE revenue > 0
)
ORDER BY budget DESC;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print("Title:", row[0], "| Budget:", row[1], "| Revenue:", row[2])
```

```python
print("Average Budget:", sum(row[1] for row in results) / len(results))
print("Average Revenue:", sum(row[2] for row in results) / len(results))
```

The query identifies movies that had a higher-than-average budget but earned lower-than-average revenue, indicating weaker financial performance despite higher investment.

```python
query = """
SELECT DISTINCT
    c.actor_name
FROM cast c
JOIN crew cr
    ON c.movie_id = cr.movie_id
WHERE cr.person_name = 'Christopher Nolan'
  AND cr.job = 'Director'
ORDER BY c.actor_name;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row[0])
```

```python
query = """
SELECT
    g.genre_name,
    AVG(m.vote_average) AS average_rating,
    COUNT(DISTINCT m.movie_id) AS movie_count
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
GROUP BY g.genre_id, g.genre_name
HAVING COUNT(DISTINCT m.movie_id) >= 20
ORDER BY average_rating DESC
LIMIT 1;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```

The query identifies the highest-rated genre among genres with at least 20 movies, giving a more reliable comparison by excluding genres with very few movies.

```python
query = """
SELECT
    m.title,
    COUNT(mk.keyword_id) AS keyword_count
FROM movies m
JOIN movie_keywords mk
    ON m.movie_id = mk.movie_id
GROUP BY m.movie_id, m.title
ORDER BY keyword_count DESC
LIMIT 10;
"""

cursor.execute(query)

results = cursor.fetchall()

for row in results:
    print(row)
```
